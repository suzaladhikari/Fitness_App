```python
### Libraries we will be using 
import pandas as pd 
import numpy as np 
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
```

```python
### Data we will be using 
data = pd.read_csv('../../Datasets/predictor.csv')
data.head(5)
```

### Train Test Split

```python
SEED = 133
X = data.drop('Calories_Burned', axis = 1) ## Columns that we will be using to predict
y = data['Calories_Burned'] ## Column we will be prediciting
X_train, X_test, y_train ,y_test = train_test_split(X,y, test_size = 0.2, random_state=SEED) 
data.drop(columns=['Session_Duration (hours)'], axis = 1)
```

### Since our data is cleaned with proper data types and has been splitted into train and test data we can use this information to train different regressor models: including LinearRegressor, RandomForestRegressor, XGBoostRegressor and many more 

---

### Model 1: LinearRegressor

```python
### Since there are no missing values in the dataset we will proceed to the pipeline

steps = [('scalar', StandardScaler()), ('linear_model', LinearRegression())]
pipeline = Pipeline(steps)
pipeline.fit(X_train, y_train)
pipeline.predict(X_test)
y_train_pred = pipeline.predict(X_train)
y_test_pred = pipeline.predict(X_test)

r2_train = r2_score(y_train, y_train_pred) # 1.0 meaning model handles all the variance in the training data properly
r2_test = r2_score(y_test, y_test_pred) #1.0 meaning model handles all the variance properly

mse_train = mean_squared_error(y_train,y_train_pred) ## ~0 which means it has really tiny error
mse_test = mean_absolute_error(y_test, y_test_pred) ## ~0 which means it has really tiny error

### How good the model is determined by r2_trained and r2_test and mse which is basically the loss function 
```
