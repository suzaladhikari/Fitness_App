### In this notebook we will be doing feature engineering for the features or the columns that we will be using for the predictor model!



```python
### Libraries we will be using
import pandas as pd 
```

```python
### Dataset we will be using 
data = pd.read_csv('../../Datasets/for_analytics.csv')
data.head(5)
print(data.columns)
```

### Since we are predicting the calories burned, we will be creating a dataset that has both dependent variable and independent varibale and also makes sure that the data types is in the same format as needed!

### The dependent varibale is Calories_Burned or we will be predicing the total calories that can be burnt ! 

```python
### Columns to drop ! 
columns_to_drop = [
    'Name of Exercise',
    'meal_name',
    'expected_burn',
    'Burns Calories (per 30 min)',
    'Burns Calories (per 30 min)_bc',
    'Burns_Calories_Bin',
    'cooking_method',
    'prep_time_min',
    'cook_time_min',
    'rating', 
    'Benefit',
    'meal_type',
    'diet_type',
    'Workout',
    'Equipment Needed',
    'Body Part',
    'Target Muscle Group',
    'Type of Muscle'
    
    
]
data = data.drop(columns = columns_to_drop, axis = 1)
```

### Lets check the dtypes of the independent columns and lets do the feature engineering 

```python
independent_columns = data.drop('Calories_Burned', axis = 1)
independent_columns.dtypes
```

### Since most of them are float, what we can do is check the categorical columns and then do the feature engineering in them 

```python
columns = data.select_dtypes('object')
data['Gender'] = data['Gender'].map({'Female':0, 'Male':1}) ## Converted the Female to 0 and Male to 1
### Since the data columns Workout_Type and Difficulty_Type has more than two different types we are doing one hot encoding
columns_dummied = ['Workout_Type', 'Difficulty Level']
data = pd.get_dummies(data, columns=columns_dummied, dtype= 'int')
```

### Since most of the feature engineering part has been done ! lets check if the data is ready for the machine learning model

```python
print(data.select_dtypes('object'))
```

### As we can see that there are no such null values, and there is no any such categorical column which can help us select the data as the best for training the model 

```python
data.to_csv('../../Datasets/predictor.csv', index = False)
```
